```python
import json
import pandas as pd

# ──────────────────────────────────────────────
# 1) scores.json 경로만 본인 환경에 맞게 수정하세요
SCORES_PATH = r"D:\kpop_app\flask-server\static\data\9534d99982174d68b0f842c5a06a7954\dancer_kp\scores.json"
threshold   = 0.8
# ──────────────────────────────────────────────

# 2) JSON 로드
with open(SCORES_PATH, 'r', encoding='utf-8') as f:
    data = json.load(f)

# 3) frame_scores는 {"0":0.75, "1":0.72, …} 형태이므로 items()로 DataFrame 생성
df_frame = pd.DataFrame(
    data['frame_scores'].items(),
    columns=['frame_idx_str', 'score']
)
# 문자열 키 → 정수로 변환, 정렬
df_frame['frame_idx'] = df_frame['frame_idx_str'].astype(int)
df_frame = df_frame.drop(columns='frame_idx_str') \
                   .sort_values('frame_idx') \
                   .reset_index(drop=True)

# 4) second_scores 도 같은 방식으로 처리
df_sec = pd.DataFrame(
    data['second_scores'].items(),
    columns=['sec_idx_str', 'avg_score']
)
df_sec['sec_idx'] = df_sec['sec_idx_str'].astype(int)
df_sec = df_sec.drop(columns='sec_idx_str') \
               .sort_values('sec_idx') \
               .reset_index(drop=True)

# 5) threshold 미만 필터링
df_frame_low = df_frame[df_frame['score'] < threshold]
df_sec_low   = df_sec[  df_sec['avg_score'] < threshold]

# 6) 결과 출력
print(f"▶ Frame별 점수 (총 {len(df_frame)}개 중 {len(df_frame_low)}개가 {threshold} 미만)")
print(df_frame_low.head(20).to_string(index=False))

print(f"\n▶ 초별 평균 점수 (총 {len(df_sec)}개 중 {len(df_sec_low)}개가 {threshold} 미만)")
print(df_sec_low.head(20).to_string(index=False))
```
